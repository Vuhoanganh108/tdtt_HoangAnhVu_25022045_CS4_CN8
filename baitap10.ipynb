```python
#W9A1
def merge(a,b,kq):
    while True:
        if a == []:
            return kq + b
        elif b == []:
            return kq + a
        if a[0] < b[0]:
            kq.append(a[0])
            a.remove(a[0])
        else:
            kq.append(b[0])
            b.remove(b[0])
if __name__ == '__main__':
    a = list(map(int,input().split()))
    b = list(map(int,input().split()))
    kq = []
    print(*merge(a,b,kq))
```

```python
#W9A2
def find_num(arr,k,lst):
    while True:
        if len(lst) == k:
            return lst[-1]
        if len(arr) == 2 and len(arr[0]) == 0:
            lst += arr[1]
            return lst[k - 1]
        if len(arr[0]) == 0:
            arr.pop(0)
        if arr[0][0] < arr[1][0]:
            lst.append(arr[0][0])
            arr[0].pop(0)
        else:
            lst.append(arr[1][0])
            arr[1].pop(0)
if __name__ == '__main__':
    arr  = []
    n = int(input('nhap so hang cua ma tran:'))
    for i in range(n):
        a = list(map(int,input().split()))
        arr.append(a)
    k = int(input())
    lst = []
    print(find_num(arr,k,lst))
```

```python
#W9A3
def swap(n):
    min = n[0]
    for i in range(len(n)):
        if int(n[i]) <= int(min):
            min = n[i]
            vi_tri = i
    if min == n[0]:
        n[-1],n[-2] = n[-2],n[-1]
    else:
        n[0],n[vi_tri] = n[vi_tri],n[0]
        if n[0] == '0': n.remove(n[0])
    return n
if __name__ == '__main__':
    n = input()
    n = list(n)
    m = swap(n)
    print(''.join(m))
```

```python
#W9A4
def check(a,b):
    id = 0
    cnt = 0
    for i in range(len(a)):

        if id != len(b) and a[i] == b[id] or a[i].upper() == b[id]:
              id += 1
              cnt += 1
        elif a[i] == a[i].upper() :
          cnt += 1
        if id == len(b) and cnt == len(b) :
            return 1
if __name__ == '__main__':
    n = int(input())
    while n > 0 :
        a = input()
        b = input()
        if check(a,b) == 1:
            print('YES')
        else: print('NO')
        n -= 1
```

```python
#W9A9
def Median(nums):
    a, b, c, d, e = nums[0], nums[1], nums[2], nums[3], nums[4]

    if a > b: a, b = b, a

    if c > d: c, d = d, c

    if a > c:
        a, c = c, a
        b, d = d, b

    if b > e:
      b, e = e, b

    if b > c:
        return b if b < d else d
    else:
        return c if c < e else e
a=list(map(int,input().split()))
print(Median(a))
```

```python
#W9A10
import bisect

def findRadius(houses, heaters):
    houses.sort()
    heaters.sort()

    max_radius = 0
    n = len(heaters)

    for house in houses:
        idx = bisect.bisect_left(heaters, house)
        dist_left = float('inf')
        dist_right = float('inf')
        if idx < n:
            dist_right = heaters[idx] - house
        if idx > 0:
            dist_left = house - heaters[idx - 1]
        shortest_to_this_house = min(dist_left, dist_right)
        max_radius = max(max_radius, shortest_to_this_house)

    return max_radius
if __name__ == "__main__":
    house = list(map(int,input().split()))
    heat = list(map(int,input().split()))
    print(findRadius(house,heat))
```
