```python
#W8A1
class Cylinder():
    def __init__(self,r,h):
        self.r = r
        self.h = h
    def dien_tich_be_mat(self):
        pi = 3.14
        return 2 * pi * self.r * (self.h + self.r)
    def the_tich(self):
        pi = 3.14
        return pi * (self.r * self.r) * self.h
r,h = map(float,input().split())
my_cylinder = Cylinder(r,h)
S = my_cylinder.dien_tich_be_mat()
V = my_cylinder.the_tich()
print(f'{S:.2f} {V:.2f}')
```

```python
#W8A2
class Date:
    def __init__(self,dd,mm,yyyy):
        self.dd = dd
        self.mm = mm
        self.yyyy = yyyy
    def is_leap(self):
        if (self.yyyy % 4 == 0 and self.yyyy % 100 != 0) or self.yyyy % 400 == 0:
            return True
    def ngay_trong_thang(self):
        if self.mm in [1,3,5,7,8,10,12] :
            return 31
        elif self.mm in [4,6,9,11]:
            return 30
        else :
            return 29 if self.is_leap() else 28
    def check(self):
        if self.dd < 1 or self.mm < 1 or self.mm > 12:
            return False
        elif self.dd > self.ngay_trong_thang():
            return False
    def ngay_ke_tiep(self):
        self.dd += 1
        if self.dd > self.ngay_trong_thang():
            self.dd = 1
            self.mm += 1
            if self.mm > 12:
                self.mm = 1
                self.yyyy += 1
        return f'{self.dd:02d}/{self.mm:02d}/{self.yyyy}'
a = input().split('/')
dd = int(a[0])
mm = int(a[1])
yyyy = int(a[2])
my_date = Date(dd,mm,yyyy)
if my_date.check() == 0 :
    print('INVALID')
else:
    print(my_date.ngay_ke_tiep())
```

```python
#W8A3
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def check(self):
        if self.x == 0 and self.y == 0:
            return 'goc toa do'
        elif self.x == 0 and self.y != 0:
            return 'truc tung'
        elif self.x != 0 and self.y == 0:
            return 'truc hoanh'
    def khoang_cach(self):
        return (self.x ** 2 + self.y ** 2) ** 0.5
M = list(map(float,input().split()))
my_point = Point(M[0],M[1])
if my_point.check() == 'goc toa do':
    print('M la goc toa do')
elif my_point.check() == 'truc hoanh':
    print('M nam tren truc hoanh')
elif my_point.check() == 'truc tung':
    print('M nam tren truc tung')
print(f'{my_point.khoang_cach():.2f}')
```

```python
#W8A5
class ShoppingCart :
    def __init__(self):
        self.cart = []
    def add(self,name,price):
        item = {'name':name,'price':price}
        self.cart.append(item)
        print(f'Da them {name} vao gio hang')
    def remove(self,name):
        for item in self.cart:
            if item['name'] == name :
                self.cart.remove(item)
                return f"da xoa {name} khoi gio hang"
        return f"khong tim thay {name} trong gio hang"
    def check(self):
        if len(self.cart) == 0:
            return 'gio hang rong'
        return 'gio hang khong rong'
    def sum(self):
        s = 0
        if self.check() == 'gio hang rong':
            return s
        for item in self.cart:
            s += item['price']
        return s
    def display(self):
        if self.check() == 'gio hang rong':
            print('khong co san pham nao trong gio hang')
        for item in self.cart :
            print(f"san pham {item['name']} co gia la {item['price']}")
    def removeall(self):
        self.cart = []
        print('da xoa toan bo gio hang')
if __name__ == '__main__':
    shop = ShoppingCart()
    while True:
        print('      MENU      ')
        print('2.Them san pham moi    3.Xoa san pham')
        print('4.Kiem tra gio hang    5.Tong tien gio hang')
        print('6.Hien thi gio hang    7.Xoa toan bo gio hang')
        chon = input()
        if chon == '2':
            ten = input('nhap ten san pham:')
            gia = int(input('nhap gia san pham:'))
            shop.add(ten,gia)
        elif chon == '3':
            ten = input('nhap san pham muon xoa:')
            print(shop.remove(ten))
        elif chon == '4':
            print(shop.check())
        elif chon == '5':
            print(shop.sum())
        elif chon == '6':
            shop.display()
        elif chon == '7':
            shop.removeall()
        print('co muon thoat khoi chuong trinh khong ?')
        ok = input()
        if ok == 'yes':
            break
```

```python
#W8A6
class Rectangle:
    def __init__(self,w,h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
    def perimeter(self):
        return 2 * (self.w + self.h)
    def scale(self,k):
        self.w = self.w * k
        self.h = self.h * k
if __name__ == '__main__':
    w,h,k = map(float,input().split())
    rec = Rectangle(w,h)
    rec.scale(k)
    print(f'{rec.area():.2f}', f'{rec.perimeter():.2f}')
```

```python
#W8A7
from math import *
class Fraction :
    def __init__(self,num,den):
        if den < 0:
            num = - num
            den = - den
        common = gcd(num,den)
        self.num = num // common
        self.den = den // common
    def add(self,other):
        new_num = self.num * other.den + self.den * other.num
        new_den = self.den * other.den
        return Fraction(new_num,new_den)
    def sub(self,other):
        new_num = self.num * other.den - self.den * other.num
        new_den = self.den * other.den
        return Fraction(new_num,new_den)
    def mul(self,other):
        new_num = self.num * other.num
        new_den = self.den * other.den
        return Fraction(new_num,new_den)
    def div(self,other):
        new_num = self.num * other.den
        new_den = self.den * other.num
        return Fraction(new_num,new_den)
if __name__ == '__main__':
    a = input().split()
    t1 = int(a[0])
    m1 = int(a[1])
    op = a[2]
    t2 = int(a[3])
    m2 = int(a[4])
    s1 = Fraction(t1,m1)
    s2 = Fraction(t2,m2)
    if op == '+':
        result = s1.add(s2)
        print(f'{result.num}/{result.den}')
    elif op == '-':
        result = s1.sub(s2)
        print(f'{result.num}/{result.den}')
    elif op == '*':
        result = s1.mul(s2)
        print(f'{result.num}/{result.den}')
    elif op == '/':
        result = s1.div(s2)
        print(f'{result.num}/{result.den}')
```

```python
#W8A8
class BankAccount:
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance
    def deposit(self,x):
        self.balance += x
    def withdraw(self,x):
        if self.balance < x:
            self.balance = self.balance
        else :
            self.balance -= x
if __name__ == '__main__':
    a = input().split()
    bank = BankAccount(a[0],float(a[1]))
    n = int(input())
    while n > 0:
        giao_dich = input().split()
        if giao_dich[0] == 'DEPOSIT':
            bank.deposit(float(giao_dich[1]))
        elif giao_dich[0] == 'WITHDRAW':
            bank.withdraw(float(giao_dich[1]))
        n -= 1
    print(f'{bank.balance:.2f}')
```

```python
#W8A9
class Student:
    def __init__(self,name):
        self.name = name
        self.sum = 0
        self.cnt = 0
    def add(self,subject,score):
        self.score = score
    def tong(self):
        self.sum += self.score
        self.cnt += 1
    def avg(self):
        result = self.sum / self.cnt
        return result
    def rank(self):
        if self.avg() >= 8:
            return 'Excellent'
        elif self.avg() >= 6.5:
            return 'Good'
        elif self.avg() >= 5:
            return 'Average'
        else:
            return 'Poor'
if __name__ == '__main__':
    name = input()
    n = int(input())
    hs = Student(name)
    while n > 0:
        subject_score = input().split()
        hs.add(subject_score[0],float(subject_score[1]))
        hs.tong()
        n -= 1
    print(name,f'{hs.avg():.2f}',hs.rank())
```

```python
#W8A10
class Book:
    def __init__(self,title,author,year):
        self.title = title
        self.author = author
        self.year = year
class Library:
    def __init__(self):
        self.book = []
    def add(self,book):
        self.book.append(book)
    def count_by_author(self,name):
        cnt = 0
        for book in self.book:
            if book.author == name:
                cnt += 1
        return cnt
    def find_by_year(self,year):
        cnt = 0
        for book in self.book:
            if book.year == year:
                cnt += 1
        return cnt
if __name__ == '__main__':
    n = int(input())
    lib = Library()
    while n > 0:
        thao_tac = input().split()
        if thao_tac[0] == 'ADD':
            b = thao_tac[1].split(';')
            book = Book(b[0],b[1],b[2])
            lib.add(book)
        elif thao_tac[0] == 'COUNT':
            print(lib.count_by_author(thao_tac[1]))
        elif thao_tac[0] == 'COUNTYEAR':
            print(lib.find_by_year(thao_tac[1]))
        n -= 1
```

```python
#W8A11
from math import *
class Angle :
    def __init__(self,a,b,c,d):
        self.ABx = b[0] - a[0]
        self.ABy = b[1] - a[1]
        self.ABz = b[2] - a[2]
        self.ACx = c[0] - a[0]
        self.ACy = c[1] - a[1]
        self.ACz = c[2] - a[2]
        self.BCx = c[0] - b[0]
        self.BCy = c[1] - b[1]
        self.BCz = c[2] - b[2]
        self.BDx = d[0] - b[0]
        self.BDy = d[1] - b[1]
        self.BDz = d[2] - b[2]
    def vector(self):
        self.ABC = [self.ABy * self.ACz - self.ABz * self.ACy, self.ABz * self.ACx - self.ABx * self.ACz, self.ABx * self.ACy - self.ABy * self.ACx]
        self.BCD = [self.BCy * self.BDz - self.BCz * self.BDy, self.BCz * self.BDx - self.BCx * self.BDz, self.BCx * self.BDy - self.BCy * self.BDx]
    def cos(self):
        tich_vo_huong = self.ABC[0] * self.BCD[0] + self.ABC[1] * self.BCD[1] + self.ABC[2] * self.BCD[2]
        tich_do_dai = ((self.ABC[0] ** 2 + self.ABC[1] ** 2 + self.ABC[2] ** 2) ** 0.5) *((self.BCD[0] ** 2 + self.BCD[1] ** 2 + self.BCD[2] ** 2) ** 0.5)
        cos = abs(tich_vo_huong/tich_do_dai)
        return cos
if __name__ == '__main__' :
    A = list(map(float,input().split()))
    B = list(map(float,input().split()))
    C = list(map(float,input().split()))
    D = list(map(float,input().split()))
    my_angle = Angle(A,B,C,D)
    my_angle.vector()
    my_angle = acos(my_angle.cos())
    my_angle = degrees(my_angle)
    print(f'{my_angle:.2f}')
```

```python
#W8A12
class Complex:
    def __init__(self,a,b):
        self.a = a
        self.b = b
    def pro(self,other):
        self.thuc = self.a * other.a - self.b * other.b
        self.ao = self.a * other.b + self.b * other.a
if __name__ == '__main__':
    a = list(map(int,input().split()))
    b = list(map(int,input().split()))
    my_complex = Complex(a[0],a[1])
    other_complex = Complex(b[0],b[1])
    my_complex.pro(other_complex)
    print(my_complex.thuc,my_complex.ao)
```

```python
#W8A13
class Triangle:
    def __init__(self,a,b,c):
        self.a = a
        self.b = b
        self.c = c
    def getArea(self):
        if self.a + self.b <= self.c or self.b + self.c <= self.a or self.a + self.c <= self.b:
            return -1
        self.p = (self.a + self.b + self.c)/2
        return (self.p * (self.p -self.a) * (self.p - self.b) * (self.p -self.c)) ** 0.5
if __name__ == '__main__':
    my_tri = Triangle(*map(int,input().split()))
    if my_tri.getArea() == -1 :
        print('invalid')
    else:
        print(f'{my_tri.getArea():.2f}')
```
